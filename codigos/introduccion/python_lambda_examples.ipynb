# Python `lambda` function examples

A `lambda` creates a small, anonymous function from a single expression. It is useful when a short function is needed briefly, such as a sorting key or a callback. For longer logic, a regular `def` function is usually clearer.

## 1. Basic syntax

The parameters come before the colon; the expression after it is the return value.

```python
double = lambda number: number * 2
double(5)
```

A lambda can take more than one argument:

```python
add = lambda first, second: first + second
add(3, 4)
```

## 2. Sorting with a key

The `key` function tells `sorted` which value to use when ordering each item. This example sorts names by their length.

```python
names = ["Charlotte", "Bo", "Amir", "Zoe"]
sorted(names, key=lambda name: len(name))
```

## 3. Transforming values with `map`

`map` applies a function to every item. Wrapping it in `list` makes the results easy to display.

```python
prices = [10, 20, 35]
prices_with_tax = list(map(lambda price: round(price * 1.10, 2), prices))
prices_with_tax
```

## 4. Selecting values with `filter`

`filter` keeps items for which the function returns `True`.

```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = list(filter(lambda number: number % 2 == 0, numbers))
even_numbers
```

## 5. A conditional expression

A lambda may use a conditional expression (sometimes called a ternary expression), but it still must be one expression.

```python
label_number = lambda number: "even" if number % 2 == 0 else "odd"
[label_number(number) for number in range(5)]
```

## 6. Lambda compared with `def`

These two functions do the same thing. Use `def` when giving a function a meaningful name or writing multiple steps; use `lambda` for short, local operations.

```python
def square(value):
    return value ** 2

square_lambda = lambda value: value ** 2

square(6), square_lambda(6)
```

## 7. Building a lambda with a fixed parameter

A regular function can take a setting (such as a multiplier) and return a lambda. The returned lambda remembers that setting, while accepting a new value each time you call it. This is called a closure.

```python
def make_multiplier(factor):
    return lambda number: number * factor

double = make_multiplier(2)  # factor is fixed at 2
triple = make_multiplier(3)  # factor is fixed at 3

double(10), triple(10)
```

## 8. Keep one argument variable and fix the others

If `f` normally takes `x`, `y`, and `z`, make a new one-argument function by writing a lambda that passes the chosen fixed values for `y` and `z`. Define those values first; afterward, only `x` needs to be supplied.

```python
def f(x, y, z):
    return x + y * z

fixed_y = 10
fixed_z = 3

f_with_fixed_yz = lambda x: f(x, fixed_y, fixed_z)

f_with_fixed_yz(2)  # same as f(2, 10, 3), which returns 32
```

## Quick guidelines

- A lambda body can contain only one expression (not statements such as `return` or `for`).
- Lambdas are handy for brief `key`, `map`, and `filter` functions.
- Prefer a named `def` if the logic is complex or will be reused.
- In many cases, a list comprehension is more readable than `map` or `filter`.
